```python
# Create small sample image folders for directory-based CNN demonstrations (Experiments 41, 43 and 46).
from PIL import Image, ImageDraw
import numpy as np, os, shutil
def make_demo_folder(root, class_names=('class_0','class_1'), n=8, size=64):
    if os.path.exists(root): shutil.rmtree(root)
    for ci, name in enumerate(class_names):
        folder=os.path.join(root,name); os.makedirs(folder,exist_ok=True)
        for i in range(n):
            arr=np.zeros((size,size,3),dtype=np.uint8)
            arr[:,:,ci]=np.random.randint(80,220,(size,size),dtype=np.uint8)
            arr[:,:,2]=np.random.randint(0,80,(size,size),dtype=np.uint8)
            Image.fromarray(arr).save(os.path.join(folder,f'{i}.jpg'))
make_demo_folder('dataset/train',('cats','dogs'),n=8)
make_demo_folder('dataset/validation',('cats','dogs'),n=4)
make_demo_folder('medical/train',('normal','abnormal'),n=8)
make_demo_folder('medical/test',('normal','abnormal'),n=4)
make_demo_folder('leaves/train',('healthy','diseased'),n=8)
make_demo_folder('leaves/validation',('healthy','diseased'),n=4)
print('Demo image datasets created.')
```

## Experiment 30:

```python
import numpy as np
# One-neuron sigmoid network; binary cross-entropy gradient
x=np.array([1.0, 0.5]); y=1.0; w=np.array([0.2,-0.1]); b=0.0; lr=0.1
sig=lambda z: 1/(1+np.exp(-z))
p=sig(x@w+b); loss=-(y*np.log(p)+(1-y)*np.log(1-p))
delta=p-y; dw=delta*x; db=delta
w-=lr*dw; b-=lr*db
p2=sig(x@w+b); loss2=-(y*np.log(p2)+(1-y)*np.log(1-p2))
print('Loss before:', round(loss,4)); print('Updated weights:', w); print('Loss after:', round(loss2,4))
```
